````python
def safe_list(x):
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return []
    return x if isinstance(x, list) else []

df['annotatorPolitics'] = df['annotatorPolitics'].apply(safe_list)
df['traditionalism'] = df['traditionalism'].apply(safe_list)
````

````python
def bucket_politics(val):
    try:
        v = float(val)
        if v <= -0.75: return "far left"
        elif -0.75 < v <= -0.25: return "left"
        elif -0.25 < v <= 0.25: return "center"
        elif 0.25 < v <= 0.75: return "right"
        else: return "far right"
    except (ValueError, TypeError):
        return "unknown"

def bucket_traditionalism(val):
    try:
        v = int(float(val))
        if v in [1, 2]: return "low"
        elif v == 3: return "medium"
        elif v in [4, 5]: return "high"
        return "unknown"
    except (ValueError, TypeError):
        return "unknown"

# Apply the buckets to the list elements
df['annotatorPolitics'] = df['annotatorPolitics'].apply(
    lambda x: [bucket_politics(v) for v in x]
)

df['traditionalism'] = df['traditionalism'].apply(
    lambda x: [bucket_traditionalism(v) for v in x]
)
````

````python
def plot_distribution(df, column_name):
    if column_name not in df.columns:
        print(f"Column '{column_name}' not found.")
        return

    values = df[column_name].explode().dropna().astype(str).str.lower().str.strip()

    if values.empty:
        print(f"No data found in {column_name}. Check if preprocessing ran correctly.")
        return

    col_lower = column_name.lower()
    title = ""
    order = None
    palette = "viridis" # Fallback palette

    if "gender" in col_lower:
        title = "Annotator Gender Distribution"
        palette = "Blues"
        order = values.value_counts().index.tolist()
    elif "race" in col_lower:
        title = "Annotator Race Distribution"
        palette = "Oranges"
        order = ["white", "black", "hispanic", "other"]
    elif "politic" in col_lower:
        title = "Annotator Political Orientation Distribution"
        palette = "Purples"
        order = ["far left", "left", "center", "right", "far right"]
    elif "trad" in col_lower:
        title = "Annotator Traditionalism Distribution"
        palette = "coolwarm"
        order = ["low", "medium", "high"]

    counts = values.value_counts().reindex(order, fill_value=0)

    if counts.sum() == 0:
        print(f"DEBUG: Found these unique values in data: {values.unique()}")
        print("These do not match the expected order labels.")

    sns.set(style="whitegrid")
    plt.figure(figsize=(8, 3))

    max_val = counts.max()
    plot_limit = max_val * 1.2 if max_val > 0 else 1.0

    ax = sns.barplot(
        x=counts.index,
        y=counts.values,
        hue=counts.index,
        palette=palette,
        legend=False,
        edgecolor="black"
    )

    for i, v in enumerate(counts.values):
        ax.text(i, v + (plot_limit * 0.02), f"{int(v)}",
                ha="center", va="bottom", fontsize=10, fontweight='bold')

    plt.title(title, pad=15)
    plt.ylabel("Count")
    plt.ylim(0, plot_limit)
    plt.tight_layout()
    plt.show()
````

````python
# DFU updated with normalisation
def dfu(input_data, histogram_input=True, normalised=True):
  """ The Distance From Unimodality measure
  :param: input_data: the data, by default the relative frequencies of ratings
  :param: histogram_input: False to compute rel. frequencies (ratings as input)
  :return: the DFU score
  """
  hist = input_data if histogram_input else to_hist(input_data)
  max_value = max(hist)
  pos_max = np.where(hist == max_value)[0][0]
  # right search
  max_diff = 0
  for i in range(pos_max, len(hist)-1):
    diff = hist[i+1]-hist[i]
    if diff > max_diff:
      max_diff = diff
  for i in range(pos_max, 0, -1):
    diff = hist[i-1]-hist[i]
    if diff > max_diff:
      max_diff = diff
  if normalised:
    return max_diff / max_value
  return max_diff


def to_hist(scores, bins_num=3, normed=True):
  """ Creating a normalised histogram
  :param: scores: the ratings (not necessarily discrete)
  :param: bins_num: the number of bins to create
  :param: normed: whether to normalise or not, by default true
  :return: the histogram
  """
  # not keeping the values order when bins are not created
  counts, bins = np.histogram(a=scores, bins=bins_num)
  counts_normed = counts/counts.sum()
  return counts_normed if normed else counts

SCALE10 = list(range(1,11))


def pdf(scores, scale=SCALE10):
  """ The relative frequencies of ordinal ratings.
  :param: scores: the ratings
  :param: scale: the rating categories, by default 10-point scale
  :return: the relative frequencies
  """
  # to be used when no bins are created
  freqs = Counter(scores)
  return np.array([freqs[s]/len(scores) for s in scale])


def cpdf(scores, scale=SCALE10):
  """ Cumulative relative frequencies
  :param: scores: the ratings
  :param: scale: the rating categories, by default 10-point scale
  :return: the cumultive relative frequencies
  """
  return np.cumsum(pdf(scores, scale))
````

````python
class UnimodalLearner:

    def __init__(self, train, dev, test, tox_threshold=0, text_col="text", scores_col="scores", scale=(-2,-1,0,1), algorithm=LogisticRegression):
        # loading the data, should comprise a column text, one with scores, and one with toxicity
        self.train = train
        self.test = test
        self.dev = dev
        self.tox_threshold = tox_threshold
        # binarized toxicity probability (lower is toxic)
        self.toxicity_func = lambda scores: np.round(np.mean([s<self.tox_threshold for s in scores]))
        # unimodality classification: if DFU is above T, then set to K+1, otherwise classify according the toxicity label
        self.labeller = lambda dfu, scores, T=0: "k+1" if dfu>T else "civil" if self.toxicity_func(scores)==0 else "toxic"
        self.scores_col = scores_col
        self.text_col = text_col
        for dataset in (self.train, self.dev, self.test):
            assert self.scores_col in dataset.columns
            dataset["HIST"] = dataset[self.scores_col].apply(lambda x: pdf(x, scale))
            dataset["DFU"] = dataset.HIST.apply(dfu) # --> changing this yields a baseline
            dataset["toxicity"] = dataset[self.scores_col].apply(self.toxicity_func)
        self.weights = {c:train.toxicity.value_counts()[c] for c in (0, 1)}
        self.thresholds = (0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1, 0)
        self.baseline_results = {"unimodal":{}, "multimodal":{}}
        self.algorithm = algorithm

    def to_np(self, dataset):
      return np.concatenate(dataset.to_numpy()).reshape(dataset.shape[0], 768) # <-- assuming BERT weights

    def pvalue(self, th=0, ssize=100, reps=range(20), averaging="macro"):
      '''
      The class reduction hypothesis: Reduce the K+1 class and compare to the
      K-class baseline
      :param th: the unimodality threshold, above this the item is said K+1
      :param ssize: the sample size
      :param reps: the repetitions of the sampling
      :param averaging: macro/micro averaging of F1
      :return: P value --> lower than 5% means the baseline lost
      '''
      model_results, baseline_results = [], []
      for _ in reps:
        e = self.test[self.test.DFU == 0].sample(ssize)
        probs = self.models[th].predict_proba(self.to_np(e["bert"]))[:, ::2] # ignoring K+1
        preds = np.argmax(probs, 1) # toxic is lower
        baseline = self.baseline.predict(self.to_np(e["bert"]))
        model_results.append(f1_score(e.toxicity.values, preds, average=averaging))
        baseline_results.append(f1_score(e.toxicity.values, baseline, average=averaging))
      statistic = scipy.stats.mannwhitneyu(baseline_results, model_results, alternative="less")
      return statistic, baseline_results, model_results


    def baseline_building(self):
        self.baseline=self.algorithm(random_state=2046, class_weight=self.weights)
        if self.algorithm is LogisticRegression:
          self.baseline.max_iter = 100000
        self.baseline.fit(self.to_np(self.train.bert), self.train.toxicity.values)
        for name, e in (("unimodal", self.dev[self.dev.DFU==0]), ("multimodal", self.dev[self.dev.DFU>0])):
            X_test = self.to_np(e["bert"])
            preds = self.baseline.predict(X_test)
            probs = self.baseline.predict_proba(X_test)
            self.baseline_results[name]["P"] = precision_score(e.toxicity.values, preds) # for toxic class
            self.baseline_results[name]["R"] = recall_score(e.toxicity.values, preds) # for toxic class
            self.baseline_results[name]["F1"] = f1_score(e.toxicity.values, preds, average='macro')
            self.baseline_results[name]["F1m"] = f1_score(e.toxicity.values, preds, average='micro')
            self.baseline_results[name]["AP"] = average_precision_score(e.toxicity.values, probs[:, 1])
            self.baseline_results[name]["AUC"] = roc_auc_score(e.toxicity.values, probs[:, 1])
            self.baseline_results[name]["support"] = len(preds)



    def baseline_tuning(self):
        self.baseline=self.algorithm(random_state=2046, class_weight=self.weights)
        if self.algorithm is LogisticRegression:
          self.baseline.max_iter = 100000
        self.baseline.fit(self.to_np(self.train.bert), self.train.toxicity.values)
        probs = self.baseline.predict_proba(self.to_np(self.dev["bert"]))
        gold = self.dev.toxicity.values
        scores = {}
        for thr in (.1, .2, .3, .4, .5, .6, .7, .8, .9):
          lbls = probs[:,1] > thr
          scores[thr] = f1_score(gold, lbls, average='macro')
        return scores


    def reduced_baseline_building(self):
        self.baseline=self.algorithm(random_state=2046, class_weight=self.weights)
        if self.algorithm is LogisticRegression:
          self.baseline.max_iter = 1000000
        train = self.train[self.train.DFU==0]
        print(f'Removed {self.train.shape[0]-train.shape[0]} instances from the {self.train.shape[0]} ones...')
        self.baseline.fit(self.to_np(train.bert), train.toxicity.values)
        preds = self.baseline.predict(self.to_np(self.test["bert"]))
        gold = self.test.toxicity.values
        return classification_report(gold, preds)



    def hypothesis_building(self, algorithm=LogisticRegression):
        self.models = {}
        for th in self.thresholds:
            self.train["target"] = self.train.apply(lambda x: self.labeller(x.DFU, x[self.scores_col], th), 1)
            self.dev["target"] = self.dev.apply(lambda x: self.labeller(x.DFU, x[self.scores_col], th), 1)
            self.test["target"] = self.test.apply(lambda x: self.labeller(x.DFU, x[self.scores_col], th), 1)

            # update the class weights for each threshold
            weight = self.train.target.value_counts()
            weights = {c:weight[c] for c in ("civil", "k+1", "toxic")}
            model = self.algorithm(random_state=2046, class_weight=weights)
            if self.algorithm is LogisticRegression:
              model.max_iter = 100000

            model.fit(self.to_np(self.train.bert), self.train.target.values)
            self.models[th] = model

    def hypothesis_testing(self, ssize=500, iterations=10, unimodality=True):
        '''
        F1 of the baseline and the K+1 model per unimodality threshold
        '''
        y = []; method_name=[]; x=[];
        for th in self.thresholds:
            for _ in range(iterations):
                condition = (self.dev.DFU == 0) if unimodality else (self.dev.DFU>0)
                e = self.dev[condition].sample(ssize)
                probs = self.models[th].predict_proba(self.to_np(e["bert"]))
                probs = probs[:, ::2] # ignoring K+1
                preds = np.argmax(probs, 1)
                y.append(precision_score(e.toxicity.values, preds)); method_name.append("Precision"); x.append(th)
                y.append(recall_score(e.toxicity.values, preds)); method_name.append("Recall");x.append(th)
                y.append(f1_score(e.toxicity.values, preds, average='macro')); method_name.append("F1-macro");x.append(th)
        results = pd.DataFrame({"Score":y, "T":x, "Measure":method_name})
        ax = sns.lineplot(data=results, x="T", y="Score", hue="Measure")
        sns.despine(left=True, bottom=True);
        ax.plot(results["T"].values, [self.baseline_results['unimodal' if unimodality else 'multimodal']["F1"]]*results.shape[0], "g*");
        return ax
````

````python
def hist_counts(ratings):
    c = Counter(ratings)
    xs = [1, 2, 3, 4, 5]
    ys = [c.get(x, 0) for x in xs]
    return xs, ys


def plot_random_au_example(df, attr="annotatorGender",
                           left_label="man", right_label="woman",
                           seed=None):
    seaborn_setup()

    if seed is not None:
        random.seed(seed)

    row = df.sample(1, random_state=seed).iloc[0]
    scores = row["toyou"]
    groups = row[attr]

    xs_all, ys_all = hist_counts(scores)

    left_sub  = [s for s, g in zip(scores, groups) if g == left_label]
    right_sub = [s for s, g in zip(scores, groups) if g == right_label]

    xs_l, ys_l = hist_counts(left_sub)
    xs_r, ys_r = hist_counts(right_sub)

    fig, axes = plt.subplots(2, 1, figsize=(5, 6), sharey=True)

    # top panel: overall — single color, no hatch
    bars_all = axes[0].bar(
        xs_all, ys_all,
        color=COLORBLIND_PALETTE[2],  # sky blue
        edgecolor="black",
    )

    axes[0].set_title("Overall distribution (polarized)", fontsize=13, pad=4)
    axes[0].set_xticks([1, 2, 3, 4, 5])
    axes[0].xaxis.label.set_visible(False)
    axes[0].yaxis.label.set_visible(False)
    axes[0].tick_params(labelsize=10)

    # bottom panel: subgroups — unchanged
    width = 0.35
    bars_l = axes[1].bar(
        np.array(xs_l) - width / 2, ys_l, width=width,
        color=COLORBLIND_PALETTE[0], edgecolor="black", label=left_label,
    )
    bars_r = axes[1].bar(
        np.array(xs_r) + width / 2, ys_r, width=width,
        color=COLORBLIND_PALETTE[1], edgecolor="black", label=right_label,
    )
    for bar in bars_l:
        bar.set_hatch(HATCHES[0])
    for bar in bars_r:
        bar.set_hatch(HATCHES[1])

    axes[1].set_title("Subgroup distributions (Unimodal)", fontsize=13, pad=4)
    axes[1].set_xticks([1, 2, 3, 4, 5])
    axes[1].xaxis.label.set_visible(False)
    axes[1].yaxis.label.set_visible(False)
    axes[1].tick_params(labelsize=10)
    axes[1].legend(fontsize=11, frameon=False)

    plt.tight_layout(pad=0.6, h_pad=1.4)
    plt.show()

    return row
````

````python
def extract_examples_with_means(df, dim_col, toxcol="toyou", threshold=3, drop_na=False):
    examples = []
    all_groups = set()

    for _, row in df.iterrows():
        entry = {"tweet": row["tweet"]}
        for grp in set(row[dim_col]):
            scores = [row[toxcol][i] for i, g in enumerate(row[dim_col]) if g == grp]

            if not scores:
                label, mean = "NA", None
            else:
                labels = ["toxic" if s >= threshold else "civil" for s in scores]
                counts = Counter(labels)
                if len(counts) == 2 and counts["toxic"] == counts["civil"]:
                    label = "tie"
                else:
                    label = max(counts, key=counts.get)
                mean = np.mean(scores)

            entry[f"{grp}_label"] = label
            entry[f"{grp}_mean"] = mean
            all_groups.add(grp)
        examples.append(entry)

    result = pd.DataFrame(examples)
    if drop_na:
      result=result.dropna(axis=1, how='any')
    return result
````

````python
def compute_AU_for_row_threshold(row, theta, toxcol="toyou", age_col="annotatorAge"):
    ages = row[age_col]
    scores = row[toxcol]

    if ages is None or scores is None or len(ages) <= 2:
        return None

    A1 = [s for s, g in zip(scores, ages) if g > theta]
    A2 = [s for s, g in zip(scores, ages) if g <= theta]

    if len(A1) <= 1 or len(A2) <= 1:
        return None

    dfu_A  = dfu(pdf(scores, scale))
    dfu_A1 = dfu(pdf(A1, scale))
    dfu_A2 = dfu(pdf(A2, scale))

    return dfu_A - max(dfu_A1, dfu_A2)
````

````python
def aggregate_AU_by_threshold(df, toxcol="toyou", age_col="annotatorAge"):
    all_ages = sorted({a for ages in df[age_col].dropna() for a in ages if pd.notna(a)})
    results = []

    for theta in all_ages:
        AU_values = df.apply(lambda r: compute_AU_for_row_threshold(r, theta, toxcol, age_col), axis=1)
        AU_values = AU_values.dropna().tolist()
        if AU_values:
            results.append({"theta": int(theta), "sum_AU": sum(AU_values),
                            "mean_AU": np.mean(AU_values), "count": len(AU_values)})

    return pd.DataFrame(results).sort_values("mean_AU", ascending=False)

# Threshold selection must use all applicable polarized texts, not the prior
# race-specific shortlist (which was left over from an earlier cell).
df_age_global = aggregate_AU_by_threshold(df[df["DFU"] > 0])
if df_age_global.empty:
    raise ValueError("No age thresholds have two or more ratings on both sides.")
best_theta = df_age_global.iloc[0]["theta"]
best_meanAU = df_age_global.iloc[0]["mean_AU"]
print(f"Best threshold across polarized texts: {best_theta}, with Mean AU = {best_meanAU:.3f}")
````

````python
def compute_AU_for_both_sides(row, theta, toxcol="toyou", age_col="annotatorAge"):
    ages = row[age_col]
    scores = row[toxcol]

    if ages is None or scores is None or len(ages) <= 2:
        return None

    A1 = [s for s, g in zip(scores, ages) if g > theta]
    A2 = [s for s, g in zip(scores, ages) if g <= theta]

    if len(A1) <= 1 or len(A2) <= 1:
        return None

    dfu_A  = dfu(pdf(scores, scale))
    dfu_A1 = dfu(pdf(A1, scale))
    dfu_A2 = dfu(pdf(A2, scale))

    return dfu_A - min(dfu_A1, dfu_A2)


def aggregate_AU_by_threshold_both_sides(df, toxcol="toyou", age_col="annotatorAge"):
    all_ages = sorted(set(a for ages in df[age_col].dropna() for a in ages if ages is not None))
    results = []

    for theta in all_ages:
        AU_values = df.apply(lambda r: compute_AU_for_both_sides(r, theta, toxcol, age_col), axis=1)
        AU_values = AU_values.dropna().tolist()
        if not AU_values:
            continue

        results.append({
            "theta": int(theta),
            "sum_AU": sum(AU_values),
            "mean_AU": np.mean(AU_values),
            "count": len(AU_values)
        })

    return pd.DataFrame(results).sort_values("mean_AU", ascending=False)


df_age_global = aggregate_AU_by_threshold_both_sides(df[df["DFU"] > 0])
print(df_age_global)

best_theta = df_age_global.iloc[0]["theta"]
best_sumAU = df_age_global.iloc[0]["sum_AU"]
print(f"Best threshold across dataset: {best_theta}, with ΣAU = {best_sumAU:.3f}")
````

````python
def strict_au_subset(data, group_col, allowed_groups=None, toxcol="toyou"):
    """Return polarized texts whose represented eligible groups are all unimodal.

    A group with one annotation is not a valid within-group distribution, so its
    partition is excluded.  Crucially, we test only groups represented in a text;
    requiring every possible politics category made attribution impossible with
    the dataset's four-to-six annotations per text.
    """
    rows = []
    for _, row in data[data["DFU"] > 0].iterrows():
        groups = row[group_col]
        scores = row[toxcol]
        if not isinstance(groups, list) or len(groups) != len(scores):
            continue
        valid = set(allowed_groups) if allowed_groups else set(groups)
        grouped = {g: [s for s, observed in zip(scores, groups) if observed == g]
                   for g in valid if g in groups}
        # Exclude singleton partitions and require a genuine comparison.
        if len(grouped) < 2 or any(len(values) < 2 for values in grouped.values()):
            continue
        if all(dfu(pdf(values, scale)) == 0 for values in grouped.values()):
            rows.append(row)
    return pd.DataFrame(rows, columns=data.columns)

x_pol = strict_au_subset(
    df, "annotatorPolitics",
    allowed_groups=["far left", "left", "center", "right", "far right"],
)
print(f"Found {len(x_pol)} politics-attributable texts.")
x_pol.head()
````

````python
def ensure_list_col(df, col):
    return df[col].apply(
        lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[")
        else ([x] if not isinstance(x, list) else x)
    )
````

````python
def build_final_summary(base_df, means_df, dimension):
    out_rows = []

    for i, row in base_df.iterrows():
        tweet = row["tweet"]
        dfu = row["DFU"]

        means_row = means_df.loc[means_df["tweet"] == tweet]
        if means_row.empty:
            continue
        means_row = means_row.iloc[0]

        poles = {}
        for col in means_row.index:
            if col.endswith("_label") and means_row[col] not in [None, "NA"] and pd.notna(means_row[col]):
                grp = col.replace("_label", "")
                poles[grp] = means_row[col]

        try:
            sorted_keys = sorted(poles.keys(), key=lambda x: float(x))
        except:
            sorted_keys = sorted(poles.keys())

        toxic_pole = "{" + ", ".join(f"{k}:{poles[k]}" for k in sorted_keys) + "}"

        out_rows.append({
            "tweet": tweet,
            "DFU": dfu,
            "dimension": dimension,
            "toxic_pole": toxic_pole
        })

    return pd.DataFrame(out_rows, columns=["tweet", "DFU", "dimension", "toxic_pole"])
````

````python
# -------------------- Helpers --------------------

def batched(iterable, batch_size):
    it = iter(enumerate(iterable))  # keep index
    while True:
        batch = list(islice(it, batch_size))
        if not batch:
            break
        yield batch

def make_prompt(batch):
    prompt = """
    Analyze the following tweets for toxicity.
    Return ONLY the score (1-5) and label (toxic/civil) for each tweet in this format:
    score=<1-5>, label=<toxic/civil>

    Tweets:
    """
    for i, (_, tweet) in enumerate(batch, start=1):
        prompt += f"{i}. {tweet}\n"
    return prompt

def parse_output(text):
    results = []
    for line in text.splitlines():
        line = line.strip()
        if not line:
            continue
        score_match = re.search(r'score[=:<]\s*([1-5])', line, re.IGNORECASE)
        label_match = re.search(r'label[=:<]\s*(toxic|civil)\b', line, re.IGNORECASE)
        if score_match and label_match:
            results.append({
                "score": int(score_match.group(1)),
                "label": label_match.group(1).lower()
            })
    return results
````

````python
def run_toxicity_analysis(tweets, completion_func, model_label, batch_size=10, retries=2):
    """Run batches without silently fabricating labels for malformed responses."""
    all_results = {}
    output_file = f"tweet_toxicity_scores_{model_label}.csv"
    print(f"Starting analysis for: {model_label.upper()}")

    for batch_idx, batch in enumerate(batched(tweets, batch_size), start=1):
        prompt = make_prompt(batch)
        last_error = None
        for _ in range(retries + 1):
            try:
                raw_answer = completion_func(prompt)
                batch_parsed = parse_output(raw_answer or "")
                if len(batch_parsed) == len(batch):
                    break
                last_error = ValueError(f"expected {len(batch)} results, received {len(batch_parsed)}")
            except Exception as exc:
                last_error = exc
        else:
            raise RuntimeError(
                f"{model_label}: batch {batch_idx} could not be parsed after {retries + 1} attempts: {last_error}"
            )

        for (orig_idx, tweet), parsed in zip(batch, batch_parsed):
            all_results[orig_idx] = {"batch_number": batch_idx, "tweet_index": orig_idx,
                                     "original_text": tweet, "score": parsed["score"],
                                     "label": parsed["label"]}

    result = pd.DataFrame.from_dict(all_results, orient="index").sort_index()
    result.to_csv(output_file, index=False)
    print(f"Results saved to {output_file}")
    return result
````

````python
def parse_toxic_pole(pole):
    if pd.isna(pole):
        return {}
    try:
        parsed = ast.literal_eval(pole) if isinstance(pole, str) else pole
        if isinstance(parsed, dict):
            return {str(k).strip("'\" "): str(v).strip("'\" ") for k, v in parsed.items()}
    except Exception:
        pass
    try:
        return {g.strip("'\" "): l.strip("'\" ")
                for g, l in [p.split(":") for p in pole.strip("{}").split(",")]}
    except Exception:
        return {}
````

````python
def compute_cohens_kappa(df, model_label_col, toxic_pole_col, group):
    model_labels, group_labels = [], []

    for _, row in df.iterrows():
        model_label = row.get(model_label_col)
        pole_dict = parse_toxic_pole(row.get(toxic_pole_col))
        if not pole_dict or group not in pole_dict:
            continue
        group_label = pole_dict[group]
        if pd.isna(model_label) or pd.isna(group_label):
            continue
        model_labels.append(str(model_label).lower())
        group_labels.append(str(group_label).lower())

    if len(model_labels) < 2:
        return np.nan

    try:
        return cohen_kappa_score(model_labels, group_labels)
    except Exception:
        return np.nan


label_map = {"civil": 0, "toxic": 1}

def compute_safe_pearson_scores(df, model_score_col, human_score_col, toxic_pole_col, group):
    model_scores = []
    human_scores = []

    for _, row in df.iterrows():
        pole_dict = parse_toxic_pole(row.get(toxic_pole_col))
        if not pole_dict or group not in pole_dict:
            continue

        m_raw = row.get(model_score_col)
        h_raw = pole_dict[group]

        # Convert labels to numbers using the map
        m = label_map.get(str(m_raw).lower())
        h = label_map.get(str(h_raw).lower())

        if m is None or h is None:
            continue

        model_scores.append(float(m))
        human_scores.append(float(h))

    if len(model_scores) < 2:
        return 0.0

    x = np.array(model_scores)
    y = np.array(human_scores)

    if np.std(x) == 0 or np.std(y) == 0:
        return 0.0

    try:
        r, _ = pearsonr(x, y)
        return 0.0 if np.isnan(r) or np.isinf(r) else r
    except:
        return 0.0
````

````python
def compute_agreement_tables(df, metric="kappa", label_col_suffix="_Label", toxic_pole_col="toxic_pole", human_score_col=None):
    model_label_cols = [c for c in df.columns if c.endswith(label_col_suffix)]
    if not model_label_cols:
        raise ValueError(f"No label columns found with suffix '{label_col_suffix}'")

    groups = set()
    for val in df[toxic_pole_col].dropna().unique():
        try:
            d = parse_toxic_pole(val)
            if isinstance(d, dict):
                groups.update(d.keys())
        except Exception:
            continue
    groups = sorted(groups)

    results = {}
    for label_col in model_label_cols:
        model_name = label_col[:-len(label_col_suffix)]
        row_vals = {}
        for g in groups:
            if metric == "kappa":
                row_vals[g] = compute_cohens_kappa(df, label_col, toxic_pole_col, g)
            elif metric == "pearson":
                # Ensure human_score_col is passed here
                row_vals[g] = compute_safe_pearson_scores(df, label_col, human_score_col, toxic_pole_col, g)
            else:
                raise ValueError("metric must be 'kappa' or 'pearson'")
        results[model_name] = row_vals

    return pd.DataFrame.from_dict(results, orient="index", columns=groups)
````

````python
def safe_kappa(y1, y2):
    y1, y2 = list(y1), list(y2)
    if len(set(y1)) == 1 and len(set(y2)) == 1:
        return 1.0 if y1[0] == y2[0] else 0.0
    try:
        return cohen_kappa_score(y1, y2)
    except Exception:
        return np.nan

````

````python
def compute_kappa_analysis(df, label_suffix="_Label"):
    label_cols = [col for col in df.columns if col.endswith(label_suffix)]
    model_names = [col.replace(label_suffix, "") for col in label_cols]

    overall_pairs = []

    for i in range(len(label_cols)):
        for j in range(i + 1, len(label_cols)):
            m1, m2 = label_cols[i], label_cols[j]

            pairs = df[[m1, m2]].dropna()

            if not pairs.empty:
                kappa = safe_kappa(pairs[m1], pairs[m2])
                overall_pairs.append({
                    "Model A": m1.replace(label_suffix, ""),
                    "Model B": m2.replace(label_suffix, ""),
                    "Kappa": round(kappa, 3)
                })

    df_results = pd.DataFrame(overall_pairs)

    if not df_results.empty:
        df_results = df_results.sort_values(by="Kappa", ascending=False)

    return df_results
````

````python
def run_debias(df, model_name, safe_completion_fn, label_col, score_col, output_file):
    records = []

    for i, row in tqdm(df.iterrows(), total=len(df), desc=f"Processing {model_name} tweets"):
        dimension = row["dimension"]
        tweet = row["tweet"]
        base = row.to_dict()

        if dimension in persona_map:
            personas = persona_map[dimension]
            prompt = make_persona_prompt(tweet, dimension, personas)

            # ---- 1) First attempt ----
            raw = safe_completion_fn(prompt, max_tokens=200)
            results = parse_output(raw, personas)

            # ---- 2) Detect unusable output & retry once ----
            if all(v["label"] is None and v["score"] is None for v in results.values()):
                print("\nBad JSON detected — retrying once")
                print("DIMENSION:", dimension)
                print("RAW OUTPUT:\n", raw)

                raw = safe_completion_fn(prompt, max_tokens=200)
                results = parse_output(raw, personas)

            labels_dict, scores_dict, same_label_dict, same_score_dict = {}, {}, {}, {}

            # ---- 3) Populate fields safely ----
            for persona in personas:
                r = results.get(persona, {}) or {}
                label = r.get("label")
                score = r.get("score")

                labels_dict[persona] = label
                scores_dict[persona] = score
                same_label_dict[persona] = (label == row[label_col]) if label is not None else None
                same_score_dict[persona] = (score == row[score_col]) if score is not None else None

        else:
            labels_dict = scores_dict = same_label_dict = same_score_dict = None

        base[f"{model_name}_Persona_Labels"] = labels_dict
        base[f"{model_name}_Persona_Scores"] = scores_dict
        base[f"{model_name}_Persona_SAME_AS_BEFORE_LABEL"] = same_label_dict
        base[f"{model_name}_Persona_SAME_AS_BEFORE_SCORE"] = same_score_dict

        records.append(base)

        if (i + 1) % 10 == 0:
            print(f"{model_name}: Processed {i+1}/{len(df)} tweets")

    persona_expanded = pd.DataFrame(records)
    persona_expanded.to_excel(output_file, index=False)

    print(f"\nFinished and saved to {output_file}!")
    return persona_expanded
````

````python
persona_map = {
    "gender": ["man", "woman"],
    "race": ["white", "black"],
    "politics": ["far_left", "left", "center", "right", "far_right"],
    "traditionalism": ["low", "high"],
    "age": ["below30", "above30"]
}
````

````python
# ---------- Prompt Builder ----------
def make_persona_prompt(tweet, dimension, personas):
    persona_list = "\n".join([f'- "{p}"' for p in personas])
    return f"""
You are an annotation system. Output ONLY valid JSON. Do not add explanations.

Tweet: "{tweet}"

Personas in the {dimension} dimension:
{persona_list}

For each persona, independently classify the tweet as "toxic" or "civil" and assign a score (1–5).
Each persona’s judgment must be made without considering the judgments of the other personas.

IMPORTANT:
- Treat each persona as a separate and independent annotation task.
- Do NOT ensure consistency or comparison across personas.
- Use the persona names EXACTLY as shown above.
- JSON keys MUST match these personas exactly.
- Do NOT substitute or rename personas.

Return ONLY valid JSON, for example:
{{
  "{personas[0]}": {{"label": "toxic", "score": 3}},
  "{personas[1]}": {{"label": "civil", "score": 1}}
}}
"""


def parse_output(raw, personas):
    raw = raw.strip()

    # strip code fences
    if raw.startswith("```"):
        raw = re.sub(r"^```[a-zA-Z]*\n?", "", raw)
        raw = re.sub(r"\n?```$", "", raw)

    parsed = {}
    try:
        parsed = json.loads(raw)
    except:
        match = re.search(r"\{[\s\S]*\}", raw)
        if match:
            try:
                parsed = json.loads(match.group(0))
            except:
                parsed = {}

    # normalize keys
    if isinstance(parsed, dict):
        parsed = {str(k).lower(): v for k, v in parsed.items()}
    else:
        parsed = {}

    if parsed and set(parsed.keys()) != set(personas):
        if len(parsed.keys()) == len(personas):
            print("ℹ️ Remapping unexpected persona keys:",
                  list(parsed.keys()), "→", personas)
            parsed = {
                p_expected: parsed[p_returned]
                for p_returned, p_expected in zip(parsed.keys(), personas)
            }

    results = {}

    # --- extract values safely ---
    for p in personas:
        label, score = None, None
        entry = parsed.get(p)

        if isinstance(entry, dict):
            label = entry.get("label")
            score = entry.get("score")
        elif isinstance(entry, str):
            label = entry.lower()

        # regex recovery if JSON missing
        if label is None:
            m = re.search(rf'"?{p}"?.*?(toxic|civil)', raw, re.IGNORECASE)
            if m:
                label = m.group(1).lower()

        if score is None:
            m = re.search(rf'"?{p}"?.*?([1-5])', raw)
            if m:
                score = int(m.group(1))

        results[p] = {"label": label, "score": score}

    return results



# ------------------ SAFE COMPLETION ------------------
def safe_completion(prompt, client, model="llama-3.1-8b-instant", max_tokens=128, retries=5):
    """
    Unified safe completion for any model behind an OpenAI-compatible client.
    """
    for attempt in range(retries):
        try:
            resp = client.chat.completions.create(
                model=model,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.0,
                max_tokens=max_tokens,
            )
            return resp.choices[0].message.content.strip()

        except Exception as exc:
            if attempt == retries - 1:
                raise
            wait_time = 2 ** attempt + random.random()
            print(f"Retry {attempt+1} after error: {exc} (waiting {wait_time:.1f}s)")
            time.sleep(wait_time)

    return "<FAILED>"
````

````python
def safe_completion(prompt, model="gemini-2.0-flash", max_tokens=256):
    for attempt in range(10):
        try:
            resp = client.chat.completions.create(
                model=model,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.0,
                max_tokens=max_tokens,
            )
            return resp.choices[0].message.content.strip()

        except Exception as exc:
            msg = str(exc)

            # Quota / rate limits
            if "RESOURCE_EXHAUSTED" in msg or "429" in msg:
                m = re.search(r"retry in (\d+)", msg)
                wait_time = int(m.group(1)) + 5 if m else 60
                print(f"⚠️ Quota hit, sleeping {wait_time}s...")
                time.sleep(wait_time)
                continue

            # Model overloaded
            if "503" in msg or "overloaded" in msg or "UNAVAILABLE" in msg:
                wait_time = min(60, 10 + attempt * 5)
                print(f"Model overloaded, sleeping {wait_time}s (attempt {attempt+1})...")
                time.sleep(wait_time)
                continue

            # Fallback exponential backoff
            if attempt == 9:
                raise
            wait_time = 2 ** attempt + random.random()
            print(f"Retry {attempt+1} after error: {exc} (waiting {wait_time:.1f}s)")
            time.sleep(wait_time)

    return "<FAILED>"
````

````python
def summarize_debias(df, model):
    def safe_parse(x):
        if isinstance(x, str):
            try:
                return ast.literal_eval(x)
            except Exception:
                return {}
        elif isinstance(x, dict):
            return x
        else:
            return {}

    persona_labels = df[f"{model}_Persona_Labels"].dropna().apply(safe_parse)
    persona_scores = df[f"{model}_Persona_Scores"].dropna().apply(safe_parse)

    persona_labels = pd.DataFrame(persona_labels.tolist())
    persona_scores = pd.DataFrame(persona_scores.tolist())

    avg_before = pd.to_numeric(df[f"{model}_Scores"], errors="coerce").mean()
    toxic_before = (df[f"{model}_Label"].astype(str).str.lower() == "toxic").mean() * 100

    avg_after = persona_scores.stack().mean()
    toxic_after = persona_labels.stack().astype(str).str.lower().eq("toxic").mean() * 100

    # --- Flip stats (safe handling of floats/NaN) ---
    def safe_all_true(val):
        parsed = safe_parse(val)
        if not isinstance(parsed, dict) or not parsed:
            return True  # treat empty as no flip
        return all(parsed.values())

    flips_label = 1 - df[f"{model}_Persona_SAME_AS_BEFORE_LABEL"].apply(safe_all_true).mean()
    flips_score = 1 - df[f"{model}_Persona_SAME_AS_BEFORE_SCORE"].apply(safe_all_true).mean()

    return {
        "Model": model,
        "Avg Score Before": round(avg_before, 2),
        "Avg Score After": round(avg_after, 2),
        "% Toxic Before": round(toxic_before, 1),
        "% Toxic After": round(toxic_after, 1),
        "% Label Flips": round(flips_label * 100, 1),
        "% Score Flips": round(flips_score * 100, 1)
    }

````

````python
# ---Helper parser ---
def parse_dict(x):
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return {}
    return x if isinstance(x, dict) else {}

# ---Compute Political Flips (macro for label, micro for score) ---
def compute_political_flips_multi(baseline_df, debias_df, model):
    merged = pd.merge(
        baseline_df[["tweet", "LLMs_Agree", "dimension"]],
        debias_df[["tweet", f"{model}_Persona_Labels", f"{model}_Persona_Scores"]],
        on="tweet", how="inner"
    )
    merged = merged[merged["dimension"].str.lower().isin(["politics", "political"])]

    label_flips, score_flips = [], []

    for _, row in merged.iterrows():
        base_poles = [p.strip().lower() for p in str(row["LLMs_Agree"]).split(",") if p.strip()]
        labels = {str(k).strip().lower(): str(v).lower() for k, v in parse_dict(row.get(f"{model}_Persona_Labels", {})).items()}
        scores = {str(k).strip().lower(): float(v) for k, v in parse_dict(row.get(f"{model}_Persona_Scores", {})).items() if isinstance(v, (int, float))}

        if not labels or not scores:
            continue

        valid_bases = [p for p in base_poles if p in labels]
        if not valid_bases:
            continue

        # Label flip (macro)
        base_labels = {labels[p] for p in valid_bases}
        other_labels = [v for k, v in labels.items() if k not in valid_bases]
        flip_label = int(any(l not in base_labels for l in other_labels))
        label_flips.append(flip_label)

        # Score flip (micro)
        base_scores = [scores[p] for p in valid_bases if p in scores]
        if not base_scores:
            continue
        base_score = sum(base_scores) / len(base_scores)
        persona_flips = [1 if abs(v - base_score) > 0 else 0 for k, v in scores.items() if k not in valid_bases]
        score_flips.extend(persona_flips)

    if not label_flips and not score_flips:
        return {"Model": model, "% Label Flips (politics)": 0.0, "% Score Flips (politics)": 0.0, "Samples": 0}

    return {
        "Model": model,
        "% Label Flips (politics)": round(sum(label_flips) / len(label_flips) * 100, 2) if label_flips else 0.0,
        "% Score Flips (politics)": round(sum(score_flips) / len(score_flips) * 100, 2) if score_flips else 0.0,
        "Samples": len(label_flips)
    }

# ---Compute General Debias Summary (for each model or subset) ---
def summarize_debias(df, model):
    def safe_parse(x):
        if isinstance(x, str):
            try:
                return ast.literal_eval(x)
            except Exception:
                return {}
        elif isinstance(x, dict):
            return x
        else:
            return {}

    persona_labels = df[f"{model}_Persona_Labels"].dropna().apply(safe_parse)
    persona_scores = df[f"{model}_Persona_Scores"].dropna().apply(safe_parse)
    persona_labels = pd.DataFrame(persona_labels.tolist())
    persona_scores = pd.DataFrame(persona_scores.tolist())

    avg_before = pd.to_numeric(df[f"{model}_Scores"], errors="coerce").mean()
    toxic_before = (df[f"{model}_Label"].astype(str).str.lower() == "toxic").mean() * 100
    avg_after = persona_scores.stack().mean()
    toxic_after = persona_labels.stack().astype(str).str.lower().eq("toxic").mean() * 100

    def safe_all_true(val):
        parsed = safe_parse(val)
        if not isinstance(parsed, dict) or not parsed:
            return True
        return all(parsed.values())

    flips_label = 1 - df[f"{model}_Persona_SAME_AS_BEFORE_LABEL"].apply(safe_all_true).mean()
    flips_score = 1 - df[f"{model}_Persona_SAME_AS_BEFORE_SCORE"].apply(safe_all_true).mean()

    return {
        "Model": model,
        "Avg Score Before": round(avg_before, 2),
        "Avg Score After": round(avg_after, 2),
        "% Toxic Before": round(toxic_before, 2),
        "% Toxic After": round(toxic_after, 2),
        "% Label Flips": round(flips_label * 100, 2),
        "% Score Flips": round(flips_score * 100, 2)
    }

# ---Run both analyses ---
baseline = pd.read_excel("LLMs_Agree.xlsx")

# Political flips
political_summaries = []
for model, debias_df in [
    ("ChatGpt", chat_debias),
    ("Gemini", gemini_debias),
    ("Mistral", mistral_debias),
    ("Llama", llama_debias),
    ("Deepseek", deepseek_debias),
    ("Grok", grok_debias)
]:
    political_summaries.append(compute_political_flips_multi(baseline, debias_df, model))

df_politics = pd.DataFrame(political_summaries)

# ---Build 4-dimension summary (gender, race, age, traditionalism) ---
dims = ["gender", "race", "age", "traditionalism","Age"]
dimension_summaries = []

for model, df in [
    ("ChatGpt", chat_debias),
    ("Gemini", gemini_debias),
    ("Mistral", mistral_debias),
    ("Llama", llama_debias),
    ("Deepseek", deepseek_debias),
    ("Grok", grok_debias),
]:
    for dim in dims:
        dim_df = df[df["dimension"].str.lower() == dim]
        if dim_df.empty:
            continue
        result = summarize_debias(dim_df, model)
        result["Dimension"] = dim
        result["Samples"] = len(dim_df)
        dimension_summaries.append(result)

df_dimension_summary = pd.DataFrame(dimension_summaries)

# --- Combine with politics ---
df_politics["Dimension"] = "politics"
df_politics.rename(columns={
    "% Label Flips (politics)": "% Label Flips",
    "% Score Flips (politics)": "% Score Flips"
}, inplace=True)

df_all_dims = pd.concat([df_dimension_summary, df_politics], ignore_index=True)
df_all_dims = df_all_dims.sort_values(["Model", "Dimension"]).reset_index(drop=True)

# ---Aggregate per model (averages across 5 attributes) ---
df_model_summary = (
    df_all_dims
    .groupby("Model", dropna=False)
    .agg({
        "Avg Score Before": "mean",
        "Avg Score After": "mean",
        "% Toxic Before": "mean",
        "% Toxic After": "mean",
        "% Label Flips": "mean",
        "% Score Flips": "mean",
        "Samples": "sum"
    })
    .reset_index()
    .round(2)
)

print("\n=== Final Combined Per-Model Summary (Averaged Across 5 Attributes, 2 decimals) ===")
display(df_model_summary)

````

````python
def parse_toxic_pole(pole):
    if pole is None:
        return {}
    if isinstance(pole, dict):
        return {str(k).strip(): str(v).strip() for k, v in pole.items()}
    if not isinstance(pole, str):
        return {}
    s = pole.strip()
    try:
        parsed = ast.literal_eval(s)
        if isinstance(parsed, dict):
            return {str(k).strip(): str(v).strip() for k, v in parsed.items()}
    except Exception:
        pass
    cleaned = s.strip("{} ").strip()
    if not cleaned:
        return {}
    pairs = re.findall(r'([^:,{}]+?)\s*:\s*([^,{}]+)', cleaned)
    result = {}
    for k, v in pairs:
        k_clean = k.strip().strip('\'"').strip()
        v_clean = v.strip().strip('\'"').strip()
        if k_clean:
            result[k_clean] = v_clean
    return result


def normalize_key(k):
    k = str(k).lower()
    k = re.sub(r'[\s_\-\u00A0]+', '', k)
    k = re.sub(r'[^a-z0-9]', '', k)
    return k.strip()


def binarize(label):
    label = str(label).lower().strip()
    if label == "toxic": return 1
    if label == "civil": return 0
    return None


EXCLUDED_GROUPS = {"hisp", "other"}


def compute_agreement_per_attribute(df, model, metric="kappa"):
    rows = []

    for dim, df_dim in df.groupby("dimension", sort=True):

        group_totals = Counter()
        for pole in df_dim["toxic_pole"].dropna():
            group_totals.update(parse_toxic_pole(pole).keys())

        for group in group_totals:
            if normalize_key(group) in EXCLUDED_GROUPS:
                continue

            group_norm = normalize_key(group)
            model_labels, human_labels = [], []

            for _, row in df_dim.iterrows():
                pole_dict = parse_toxic_pole(row["toxic_pole"])
                if not pole_dict:
                    continue

                human_label = None
                for k, v in pole_dict.items():
                    if normalize_key(k) == group_norm:
                        human_label = v
                        break
                if human_label is None:
                    continue
                h = binarize(human_label)
                if h is None:
                    continue

                persona_raw = row.get(f"{model}_Persona_Labels")
                if persona_raw is None or (isinstance(persona_raw, float) and np.isnan(persona_raw)):
                    continue

                if isinstance(persona_raw, str):
                    try:
                        persona_labels = ast.literal_eval(persona_raw)
                    except Exception:
                        pairs = re.findall(r'([^:,{}]+?)\s*:\s*([^,{}]+)', persona_raw)
                        persona_labels = {
                            k.strip().strip('\'"'): v.strip().strip('\'"')
                            for k, v in pairs
                        }
                elif isinstance(persona_raw, dict):
                    persona_labels = persona_raw
                else:
                    continue

                if not isinstance(persona_labels, dict):
                    continue

                persona_labels_norm = {normalize_key(k): v for k, v in persona_labels.items()}
                model_label = persona_labels_norm.get(group_norm)
                if model_label is None:
                    continue
                p = binarize(model_label)
                if p is None:
                    continue

                model_labels.append(p)
                human_labels.append(h)

            n = len(model_labels)
            score = np.nan

            if n >= 2:
                if metric == "pearson":
                    if np.std(model_labels) > 0 and np.std(human_labels) > 0:
                        r, _ = pearsonr(model_labels, human_labels)
                        score = r
                else:
                    if len(set(model_labels)) > 1 and len(set(human_labels)) > 1:
                        score = cohen_kappa_score(model_labels, human_labels)
                    else:
                        score = 0.0

            rows.append({
                "dimension": dim,
                "group":     group,
                "examples":  n,
                model:       round(score, 2) if not np.isnan(score) else np.nan
            })

    return pd.DataFrame(rows)


def compute_all_agreements(metric="kappa"):
    model_dfs = [
        ("ChatGpt",  chat_debias),
        ("Gemini",   gemini_debias),
        ("Mistral",  mistral_debias),
        ("Llama",    llama_debias),
        ("Deepseek", deepseek_debias),
        ("Grok",     grok_debias),
    ]
    numeric_cols = ["ChatGpt", "Gemini", "Mistral", "Llama", "Deepseek", "Grok"]

    summaries = []
    for model, df in model_dfs:
        summaries.append(compute_agreement_per_attribute(df, model, metric=metric))

    df_all = summaries[0][["dimension", "group", "examples", "ChatGpt"]]
    for s in summaries[1:]:
        model_col = [c for c in s.columns if c not in ["dimension", "group", "examples"]][0]
        df_all = df_all.merge(
            s[["dimension", "group", model_col]],
            on=["dimension", "group"],
            how="outer"
        )

    expected = [{"dimension": dim, "group": g} for dim, groups in persona_map.items() for g in groups]
    df_all = pd.DataFrame(expected).merge(df_all, on=["dimension", "group"], how="left")
    df_all["examples"] = df_all["examples"].fillna(0).astype(int)
    df_all[numeric_cols] = df_all[numeric_cols].round(2)
    return df_all


sensitivity_kappa   = compute_all_agreements(metric="kappa")
sensitivity_pearson = compute_all_agreements(metric="pearson")

numeric_cols = ["ChatGpt", "Gemini", "Mistral", "Llama", "Deepseek", "Grok"]
````

````python
def to_scalar_score(x):
    if isinstance(x, (int, float)):
        return x
    try:
        d = ast.literal_eval(x) if isinstance(x, str) else x
        if isinstance(d, dict):
            vals = [v for v in d.values() if v is not None]
            return sum(vals) / len(vals) if vals else None
    except:
        return None
````

````python
from collections import defaultdict, Counter
from itertools import product
EXCLUDED_GROUPS = {"hisp", "other"}

def parse_toxic_pole(pole):
    if pole is None:
        return {}
    if isinstance(pole, dict):
        return {str(k).strip(): str(v).strip() for k, v in pole.items()}
    if not isinstance(pole, str):
        return {}
    s = pole.strip()
    try:
        parsed = ast.literal_eval(s)
        if isinstance(parsed, dict):
            return {str(k).strip(): str(v).strip() for k, v in parsed.items()}
    except Exception:
        pass
    cleaned = s.strip("{} ").strip()
    if not cleaned:
        return {}
    pairs = re.findall(r'([^:,{}]+?)\s*:\s*([^,{}]+)', cleaned)
    result = {}
    for k, v in pairs:
        k_clean = k.strip().strip('\'"').strip()
        v_clean = v.strip().strip('\'"').strip()
        if k_clean:
            result[k_clean] = v_clean
    return result


def normalize_key(k):
    k = str(k).lower()
    k = re.sub(r'[\s_\-\u00A0]+', '', k)
    k = re.sub(r'[^a-z0-9]', '', k)
    return k.strip()


def binarize(label):
    label = str(label).lower().strip()
    if label == "toxic": return 1
    if label == "civil": return 0
    return None


def parse_persona_labels_from_row(row, model):
    persona_raw = row.get(f"{model}_Persona_Labels")
    if persona_raw is None or (isinstance(persona_raw, float) and np.isnan(persona_raw)):
        return {}
    if isinstance(persona_raw, dict):
        return {normalize_key(k): str(v).strip() for k, v in persona_raw.items()}
    if isinstance(persona_raw, str):
        try:
            parsed = ast.literal_eval(persona_raw)
            if isinstance(parsed, dict):
                return {normalize_key(k): str(v).strip() for k, v in parsed.items()}
        except Exception:
            pass
        pairs = re.findall(r'([^:,{}]+?)\s*:\s*([^,{}]+)', persona_raw)
        return {normalize_key(k.strip().strip('\'"')): v.strip().strip('\'"') for k, v in pairs}
    return {}


def compute_kappa_pair(df, model, sg_persona, sg_pole):
    sg_persona_norm = normalize_key(sg_persona)
    sg_pole_norm    = normalize_key(sg_pole)
    model_labels, human_labels = [], []

    for _, row in df.iterrows():
        # Human label
        pole_dict = parse_toxic_pole(row.get("toxic_pole"))
        if not pole_dict:
            continue
        human_label = None
        for k, v in pole_dict.items():
            if normalize_key(k) == sg_pole_norm:
                human_label = v
                break
        if human_label is None:
            continue
        h = binarize(human_label)
        if h is None:
            continue

        # Model label
        persona_labels = parse_persona_labels_from_row(row, model)
        model_label = persona_labels.get(sg_persona_norm)
        if model_label is None:
            continue
        p = binarize(model_label)
        if p is None:
            continue

        model_labels.append(p)
        human_labels.append(h)

    n = len(model_labels)
    if n < 2 or len(set(model_labels)) < 2 or len(set(human_labels)) < 2:
        return 0.0, n
    return round(cohen_kappa_score(human_labels, model_labels), 4), n


# --- Intra-group agreement (Table 5) ---
def compute_all_agreements(metric="kappa"):
    model_list = [
        ("ChatGpt",  chat_debias),
        ("Gemini",   gemini_debias),
        ("Mistral",  mistral_debias),
        ("Llama",    llama_debias),
        ("Deepseek", deepseek_debias),
        ("Grok",     grok_debias),
    ]
    numeric_cols = ["ChatGpt", "Gemini", "Mistral", "Llama", "Deepseek", "Grok"]

    summaries = []
    for model, df in model_list:
        rows = []
        for dim, df_dim in df.groupby("dimension", sort=True):
            group_totals = Counter()
            for pole in df_dim["toxic_pole"].dropna():
                group_totals.update(parse_toxic_pole(pole).keys())
            for group in group_totals:
                score, n = compute_kappa_pair(df_dim, model, group, group)
                rows.append({
                    "dimension": dim,
                    "group":     group,
                    "examples":  n,
                    model:       round(score, 2) if score is not None else np.nan
                })
        summaries.append(pd.DataFrame(rows))

    df_all = summaries[0][["dimension", "group", "examples", "ChatGpt"]]
    for s in summaries[1:]:
        model_col = [c for c in s.columns if c not in ["dimension", "group", "examples"]][0]
        df_all = df_all.merge(s[["dimension", "group", model_col]], on=["dimension", "group"], how="outer")

    expected = [{"dimension": dim, "group": g} for dim, groups in persona_map.items() for g in groups]
    df_all = pd.DataFrame(expected).merge(df_all, on=["dimension", "group"], how="left")
    df_all["examples"] = df_all["examples"].fillna(0).astype(int)
    df_all[numeric_cols] = df_all[numeric_cols].round(2)
    return df_all


# --- Cross-subgroup tables (Appendix) ---
def compute_cross_subgroup_tables():
    model_list = [
        ("ChatGpt",  chat_debias),
        ("Gemini",   gemini_debias),
        ("Mistral",  mistral_debias),
        ("Llama",    llama_debias),
        ("Deepseek", deepseek_debias),
        ("Grok",     grok_debias),
    ]
    model_cols = [m for m, _ in model_list]

    dim_subgroups = defaultdict(set)
    for model_col, df in model_list:
        for dim, df_dim in df.groupby("dimension"):
            for pole in df_dim["toxic_pole"].dropna():
                for k in parse_toxic_pole(pole).keys():
                  nk = normalize_key(k)
                  if nk not in EXCLUDED_GROUPS:
                      dim_subgroups[dim].add(nk)

    tables = {}
    for dim, subgroups in dim_subgroups.items():
        subgroups = sorted(subgroups)
        rows_out = []
        for sg_persona, sg_pole in product(subgroups, repeat=2):
            row_dict = {"pair": f"{sg_persona}  vs  {sg_pole}"}
            for model_col, df in model_list:
                df_dim = df[df["dimension"] == dim]
                score, n = compute_kappa_pair(df_dim, model_col, sg_persona, sg_pole)
                row_dict[model_col] = score
                row_dict[f"{model_col}_n"] = n
            rows_out.append(row_dict)
        tables[dim] = pd.DataFrame(rows_out)

    return tables, model_cols


# --- Run ---
sensitivity_kappa = compute_all_agreements()
cross_tables, model_cols = compute_cross_subgroup_tables()

numeric_cols = ["ChatGpt", "Gemini", "Mistral", "Llama", "Deepseek", "Grok"]

for dim, df in cross_tables.items():
    print(f"\n{'='*70}\n  {dim.upper()}\n{'='*70}")
    print(df[["pair"] + model_cols].to_string(index=False))
````
